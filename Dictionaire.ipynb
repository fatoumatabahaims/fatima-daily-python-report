<br><br><b><span style="color:red; font-size:30px;">6. Parcourir un dictionnaire</span></b><br><br>

```python
for cle in personne:               
    print(cle)
```

```python
for valeur in personne.values():   
    print(valeur)
```

```python
for cle, valeur in personne.items():  # clé + valeur
    print(cle, valeur)
```

<br><br><b><span style="color:red; font-size:30px;">7. Copier / Fusionner</span></b><br><br>

```python
autre = {"email": "diallo@mail.com"}
personne.update(autre) 
personne
```
